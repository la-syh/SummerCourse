```python
from bs4 import BeautifulSoup

with open('index.html', 'r', encoding = 'utf -8') as file:
    html_doc = file.read()
soup = BeautifulSoup(html_doc, 'html.parser')
```

```python
all_links = set()
for anchor in soup.find_all('a'):
    href = anchor.attrs.get('href')
    if href != '' and href != None:
        all_links.add(href)
print(all_links)
```
