```python
from bs4 import BeautifulSoup
import jieba

with open('index.html', 'r', encoding = 'utf-8') as file:
    html_doc = file.read()
soup = BeautifulSoup(html_doc, 'html5lib')

# print(soup.prettify())
```

```python
# 提取文档中所有链接
all_links = set()
for anchor in soup.find_all('a'):
    href = anchor.attrs.get('href')
    if href != '' and href != None:
        all_links.add(href)
print(all_links)
```

```python
# 抓取标题并分词
title_string = ''
if soup.title is not None:
    title_string = soup.title.string
title_words = jieba.lcut(title_string)
print(title_words)

doctitle_words = []
for t in soup.find_all({'h1', 'h2', 'h3'}):
    if t.string is not None:
        doctitle_words += jieba.lcut(t.string)
print(doctitle_words)
```

```python
# 抓取正文并分词
body_words = []
for p in soup.find_all('p'):
    if p.string is not None:
        body_words += jieba.lcut(p.string)
print(body_words)
```
